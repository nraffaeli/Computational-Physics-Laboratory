```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams.update({
    "font.family": "serif",
    "mathtext.fontset": "dejavuserif",
    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.top": True,
    "ytick.right": True,
    "axes.linewidth": 1.5,
})
```

```python
tc = np.loadtxt("../ColdThermalization.txt")
th = np.loadtxt("../HotThermalization.txt")
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4), constrained_layout=True)

for ax in (ax1, ax2):
    ax.plot(th[:, 0], th[:, 1], color="C1", lw=1.2, label="Hot start")
    ax.plot(tc[:, 0], tc[:, 1], color="C0", lw=1.2, label="Cold start")
    ax.set_xlabel(r"$N_{\mathrm{sweeps}}$")
    ax.grid(alpha=0.3)
    ax.minorticks_on()

ax1.set_yscale("log")
ax1.set_ylabel("Action")
ax1.set_title("Thermalization")
ax1.legend(frameon=False)

plateau = np.concatenate([tc[200:, 1], th[200:, 1]])
m, s = plateau.mean(), plateau.std()
ax2.axhline(m, color="k", ls="--", lw=0.8)
ax2.set_ylim(m - 5*s, m + 5*s)
ax2.set_title("Zoom on plateau")

plt.show()
```
